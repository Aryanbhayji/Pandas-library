```python
##pd.crosstab() creates a frequency table that shows how often different combinations of categorical values occur.

import pandas as pd

data = {
    "Name": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Gender": [
        "Male", "Female", "Male", "Female",
        "Male", "Female", "Male", "Female"
    ],
    "Department": [
        "CSE", "CSE", "ECE", "ECE",
        "CSE", "ECE", "CSE", "ECE"
    ],
    "Result": [
        "Pass", "Pass", "Fail", "Pass",
        "Pass", "Fail", "Fail", "Pass"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
table = pd.crosstab(
    df["Department"],
    df["Gender"]
)

print(table)
```

```python
pd.crosstab(
    df["Department"],
    df["Gender"],
    normalize=True
) * 100
```
