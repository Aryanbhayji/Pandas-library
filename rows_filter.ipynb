```python
import pandas as pd 
```

```python
data = {
    "name":['Ram', 'Shyam','ghanshyam', 'dhanshyam','Aditi','jagdish','Raj','Simran'],
    "Age":[10,20,30,40,50,60,70,80],
    "Salary":[50000,60000,40000,80000,20000,15000,30000,24000],
    "performance score ":[85,90,78,92,88,95,80,89]

}

df = pd.DataFrame(data)

highSalary = df[df['Salary']>50000]
print('employees with salary > 50000')
print(highSalary)
```

```python
#filtering row salary >50k & age>13

data = {
    "name":['Ram', 'Shyam','ghanshyam', 'dhanshyam','Aditi','jagdish','Raj','Simran'],
    "Age":[10,20,30,40,50,60,70,80],
    "Salary":[50000,60000,40000,80000,20000,15000,30000,24000],
    "performance_score":[85,90,78,92,88,95,80,89]

}

df = pd.DataFrame(data)

filtered_row = df[(df['Age']>30) & (df['Salary']>20000)]
print(filtered_row)

#using or condition
filtered_row_2 = df[(df['Age']>35) | (df['performance_score']>90)]
print(filtered_row_2)
```
