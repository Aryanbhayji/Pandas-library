```python
import pandas as pd
```

```python
sales = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Tablet"],
    "January": [50000, 30000, 20000],
    "February": [60000, 35000, 25000],
    "March": [55000, 40000, 30000]
})

print(sales)
```

```python
#Purpose: melt() converts a DataFrame from wide format into long format.

long_sales = pd.melt(
    sales,
    id_vars=["Product"],
    var_name="Month",
    value_name="Revenue"
)

print(long_sales)
```

```python
##Parameter	Purpose
""" frame	Original DataFrame
id_vars -	Columns to keep unchanged
value_vars-	Columns to convert into rows
var_name-	Name of the new variable column
value_name-	Name of the new values column
ignore_index-	Whether to create a new index """
```

```python
pd.melt(
    sales,
    id_vars=["Product"],
    value_vars=["January", "February"],
    var_name="Month",
    value_name="Revenue"
)
```
