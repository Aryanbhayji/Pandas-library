```python
import pandas as pd
```

```python
##pivot_table() is used to summarize large datasets by arranging data into rows and columns and applying calculations such as
#  mean, sum, count, maximum, and minimum.


data = {
    "Name": ["A", "B", "C", "D", "E", "F"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"],
    "Department": ["CSE", "CSE", "ECE", "ECE", "CSE", "ECE"],
    "Marks": [85, 90, 75, 80, 95, 88]
}

df = pd.DataFrame(data)

print(df)
```

```python
table = pd.pivot_table(
    df,
    values="Marks",
    index="Department",
    aggfunc="mean"
)

print(table)
```

```python
# Average
aggfunc="mean"

# Total
aggfunc="sum"

# Number of records
aggfunc="count"

# Highest value
aggfunc="max"

# Lowest value
aggfunc="min"

# Number of unique values
aggfunc="nunique"
```

```python
pd.pivot_table(
    df,
    values="Marks",
    index="Department",
    aggfunc=["mean", "max", "min", "count"]
)
```

```python
##margins=True
##Suppose you want the overall average along with department averages.

pd.pivot_table(
    df,
    values="Marks",
    index="Department",
    aggfunc="mean",
    margins=True
)
```
