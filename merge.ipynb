```python
#Purpose: merge() combines two DataFrames using a common column or key.

import pandas as pd
import numpy as np

# Customer Dataset
customers = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105],
    "name": ["Aryan", "Rahul", "Aman", "Rohit", "Priya"],
    "city": ["Delhi", "Mumbai", "Lucknow", "Kanpur", "Jaipur"]
})

# Orders Dataset
orders = pd.DataFrame({
    "customer_id": [101, 102, 101, 103, 106],
    "order_id": [501, 502, 503, 504, 505],
    "amount": [5000, 3000, 2000, 4500, 6000]
})

print("CUSTOMERS DATA:")
print(customers)

print("\nORDERS DATA:")
print(orders)
```

```python
#Basic Merge

basic_merge = pd.merge(
    customers,
    orders,
    on="customer_id"
)

print(basic_merge)
```

```python
#Inner Join

inner_join = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="inner"
)

print(inner_join)
```

```python
#Left Join

left_join = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="left"
)

print(left_join)
```

```python
#Right Join

right_join = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="right"
)

print(right_join)
```

```python
#Outer Join

outer_join = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="outer"
)

print(outer_join)
```

```python
#Merge with Different Column Names
#Sometimes two DataFrames have different column names for the same information.

students = pd.DataFrame({
    "student_id": [1, 2, 3],
    "student_name": ["Aryan", "Rahul", "Aman"]
})

marks = pd.DataFrame({
    "id": [1, 2, 3],
    "marks": [85, 90, 78]
})

result = pd.merge(
    students,
    marks,
    left_on="student_id",
    right_on="id",
    how="inner"
)

print(result)
```

```python
#Merge Validation
#Purpose: Helps check whether the relationship between two DataFrames matches your expectations.

result = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="left",
    validate="one_to_many"
)

print(result)


"""  Other validation options:
Validation	Meaning
one_to_one	- Unique keys in both DataFrames
one_to_many	- Unique keys in left DataFrame
many_to_one	- Unique keys in right DataFrame
many_to_many- Multiple matching keys allowed  """
```
