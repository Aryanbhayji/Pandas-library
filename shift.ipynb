The shift() function moves data up or down by a specified number of rows.

It is especially useful when you want to compare the current value with the previous value.

```python
import pandas as pd

data = {
    "Day": ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"],
    "Sales": [100, 150, 120, 200, 250]
}

df = pd.DataFrame(data)

df["Previous_Day_Sales"] = df["Sales"].shift(1)

print(df)
```

```python
df["Next_Day_Sales"] = df["Sales"].shift(-1)
print(df)
```

```python
#Now let's calculate daily sales growth.

df["Previous_Sales"] = df["Sales"].shift(1)

df["Sales_Change"] = (
    df["Sales"] - df["Previous_Sales"]
)

df["Growth_Percentage"] = (
    df["Sales"].pct_change() * 100
)

print(df)
```
